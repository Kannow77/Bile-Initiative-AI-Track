```python
df['age'].median()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df.isnull().sum()
```

#Clean gender

```python
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
```

```python
df["gender"].isnull().sum()
```

#city clean

```python
df["city"].mode()
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
```

```python
df.isnull().sum()
```

#attendance pecentage

```python
df["attendance_percent"].median()
```

```python
df["attendance_percent"] = df["attendance_percent"].fillna(
    df["attendance_percent"].median()
)
```

```python
df["attendance_percent"].isnull().sum()
```

#exam score


```python
df["exam_score"].median()
```

```python
df["exam_score"] = df["exam_score"].fillna(
    df["exam_score"].median()
)
```

```python
df["exam_score"].isnull().sum()
```

```python
df.isnull().sum()
```

#duplicate

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
# Check cleaned data
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
print("Shape:", df.shape)
```
